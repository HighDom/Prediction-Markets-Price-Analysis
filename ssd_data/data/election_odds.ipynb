```python
# Electoral Vote Calculator with Per-State Probabilities

# Assigned states data
assigned_states = [
    # Solid Harris states
    {'name': 'CA', 'votes': 54, 'assigned_to': 'Harris'},
    {'name': 'NY', 'votes': 28, 'assigned_to': 'Harris'},
    {'name': 'IL', 'votes': 19, 'assigned_to': 'Harris'},
    {'name': 'WA', 'votes': 12, 'assigned_to': 'Harris'},
    {'name': 'MA', 'votes': 11, 'assigned_to': 'Harris'},
    {'name': 'MD', 'votes': 10, 'assigned_to': 'Harris'},
    {'name': 'CO', 'votes': 10, 'assigned_to': 'Harris'},
    {'name': 'CT', 'votes': 7,  'assigned_to': 'Harris'},
    {'name': 'OR', 'votes': 8,  'assigned_to': 'Harris'},
    {'name': 'NJ', 'votes': 14, 'assigned_to': 'Harris'},
    {'name': 'HI', 'votes': 4,  'assigned_to': 'Harris'},
    {'name': 'RI', 'votes': 4,  'assigned_to': 'Harris'},
    {'name': 'DE', 'votes': 3,  'assigned_to': 'Harris'},
    {'name': 'DC', 'votes': 3,  'assigned_to': 'Harris'},
    {'name': 'VT', 'votes': 3,  'assigned_to': 'Harris'},

    # Solid Trump states
    {'name': 'ID', 'votes': 4,  'assigned_to': 'Trump'},
    {'name': 'MT', 'votes': 4,  'assigned_to': 'Trump'},
    {'name': 'ND', 'votes': 3,  'assigned_to': 'Trump'},
    {'name': 'SD', 'votes': 3,  'assigned_to': 'Trump'},
    {'name': 'WY', 'votes': 3,  'assigned_to': 'Trump'},
    {'name': 'UT', 'votes': 6,  'assigned_to': 'Trump'},
    {'name': 'IA', 'votes': 6,  'assigned_to': 'Trump'},
    {'name': 'KS', 'votes': 6,  'assigned_to': 'Trump'},
    {'name': 'MO', 'votes': 10, 'assigned_to': 'Trump'},
    {'name': 'OK', 'votes': 7,  'assigned_to': 'Trump'},
    {'name': 'TX', 'votes': 40, 'assigned_to': 'Trump'},
    {'name': 'AR', 'votes': 6,  'assigned_to': 'Trump'},
    {'name': 'LA', 'votes': 8,  'assigned_to': 'Trump'},
    {'name': 'MS', 'votes': 6,  'assigned_to': 'Trump'},
    {'name': 'AL', 'votes': 9,  'assigned_to': 'Trump'},
    {'name': 'FL', 'votes': 30, 'assigned_to': 'Trump'},
    {'name': 'SC', 'votes': 9,  'assigned_to': 'Trump'},
    {'name': 'TN', 'votes': 11, 'assigned_to': 'Trump'},
    {'name': 'KY', 'votes': 8,  'assigned_to': 'Trump'},
    {'name': 'IN', 'votes': 11, 'assigned_to': 'Trump'},
    {'name': 'OH', 'votes': 17, 'assigned_to': 'Trump'},
    {'name': 'WV', 'votes': 4,  'assigned_to': 'Trump'},
    {'name': 'AK', 'votes': 3,  'assigned_to': 'Trump'},
]

# Special handling for NE and ME
# NE: Trump wins statewide (2 votes) and 2 districts; Harris wins 1 district
# ME: Harris wins statewide (2 votes) and 1 district; Trump wins 1 district
ne_votes = {'Trump': 4, 'Harris': 1}  # Total 5 votes
me_votes = {'Trump': 1, 'Harris': 3}  # Total 4 votes

# Calculate assigned electoral votes
harris_assigned_votes = sum(state['votes'] for state in assigned_states
                            if state['assigned_to'] == 'Harris') + ne_votes['Harris'] + me_votes['Harris']
trump_assigned_votes = sum(state['votes'] for state in assigned_states
                           if state['assigned_to'] == 'Trump') + ne_votes['Trump'] + me_votes['Trump']

# Swing states with provided probabilities
swing_states = [
    {'name': 'AZ', 'votes': 11, 'p_trump': 0.73},
    {'name': 'GA', 'votes': 16, 'p_trump': 0.74},
    {'name': 'NV', 'votes': 6,  'p_trump': 0.64},
    {'name': 'PA', 'votes': 19, 'p_trump': 0.62},
    {'name': 'WI', 'votes': 10, 'p_trump': 0.58},
    {'name': 'MI', 'votes': 15, 'p_trump': 0.53},
]
swing_states_new = [
    {'name': 'AZ', 'votes': 11, 'p_trump': 0.78},
    {'name': 'GA', 'votes': 16, 'p_trump': 0.67},
    {'name': 'NV', 'votes': 6,  'p_trump': 0.58},
    {'name': 'PA', 'votes': 19, 'p_trump': 0.59},
    {'name': 'WI', 'votes': 10, 'p_trump': 0.44},
    {'name': 'MI', 'votes': 15, 'p_trump': 0.40},
]

# Other unclear states (default probability of 0.5 for Trump)
other_unclear_states = [
    {'name': 'MN', 'votes': 10, 'p_trump': 0.17},
    {'name': 'NM', 'votes': 5,  'p_trump': 0.14},
    {'name': 'VA', 'votes': 13, 'p_trump': 0.18},
    {'name': 'NH', 'votes': 4,  'p_trump': 0.19},
    {'name': 'NC', 'votes': 16, 'p_trump': 0.71},
]

other_unclear_states_new = [
    {'name': 'MN', 'votes': 10, 'p_trump': 0.09},
    {'name': 'NM', 'votes': 5,  'p_trump': 0.09},
    {'name': 'VA', 'votes': 13, 'p_trump': 0.1},
    {'name': 'NH', 'votes': 4,  'p_trump': 0.20},
    {'name': 'NC', 'votes': 16, 'p_trump': 0.68},
]

# Combine all unclear states
unclear_states = swing_states_new + other_unclear_states_new

# Initialize totals
trump_electoral_votes = trump_assigned_votes
harris_electoral_votes = harris_assigned_votes

# Iterate through unclear states and calculate weighted votes
for state in unclear_states:
    votes = state['votes']
    p_trump = state['p_trump']
    p_harris = 1 - p_trump

    # Weighted electoral votes for each candidate
    trump_electoral_votes += votes * p_trump
    harris_electoral_votes += votes * p_harris

# Print results
print("Total Weighted Electoral Votes:")
print(f"Trump: {trump_electoral_votes:.2f}")
print(f"Harris: {harris_electoral_votes:.2f}")

from itertools import product
from tqdm import tqdm

n_states = len(unclear_states)
total_combinations = 2 ** n_states

harris_win_prob = 0.0
trump_win_prob = 0.0
tie_prob = 0.0

print("Calculating probabilities...")

# TQDM -> Progress bar, product -> Cartesian product of states
for outcome in tqdm(product([0, 1], repeat=n_states), total=total_combinations):
    harris_votes = harris_assigned_votes
    trump_votes = trump_assigned_votes  
    prob = 1.0

    for idx, result in enumerate(outcome):
        state = unclear_states[idx]
        votes = state['votes']
        p_trump = state['p_trump']
        p_harris = 1 - p_trump

        if result == 1:
            # Trump wins the state
            trump_votes += votes
            prob *= p_trump
        else:
            # Harris wins the state
            harris_votes += votes
            prob *= p_harris

    if harris_votes >= 270 and harris_votes > trump_votes:
        harris_win_prob += prob
    elif trump_votes >= 270 and trump_votes > harris_votes:
        trump_win_prob += prob
    elif harris_votes == trump_votes:
        tie_prob += prob

# Normalize probabilities
total_prob = harris_win_prob + trump_win_prob + tie_prob

print("\nNormalization factor:", total_prob)

harris_win_prob /= total_prob
trump_win_prob /= total_prob
tie_prob /= total_prob

# Display the results
print("\nOdds of Each Candidate Winning:")
print(f"Harris win probability: {harris_win_prob * 100:.2f}%")
print(f"Trump win probability: {trump_win_prob * 100:.2f}%")
print(f"Tie probability: {tie_prob * 100:.2f}%")
```
