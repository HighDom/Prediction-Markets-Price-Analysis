```python
# Transaction Analysis

# Import Libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

%matplotlib inline

## 1. Load Data
# Define file ranges
start, end = 50, 61  # Adjust these variables to load specific files

# Load the PositionsMerge and PositionsSplit data
positions_merge = pd.read_csv(f'positionsmerge_events/positionsmerge_{start}-{end}.csv')
positions_split = pd.read_csv(f'positionsplit_events/positionsplit_{start}-{end}.csv')

# Display the loaded data
print("PositionsMerge Data:")
display(positions_merge.head())

print("\nPositionsSplit Data:")
display(positions_split.head())

## 2. Data Preprocessing
# Ensure that the `amount` column is numeric and handle any missing values if necessary.
positions_merge['amount'] = pd.to_numeric(positions_merge['amount'], errors='coerce')
positions_split['amount'] = pd.to_numeric(positions_split['amount'], errors='coerce')

positions_merge = positions_merge.dropna(subset=['amount'])
positions_split = positions_split.dropna(subset=['amount'])

positions_merge['conditionId'] = positions_merge['conditionId'].astype(str)
positions_split['conditionId'] = positions_split['conditionId'].astype(str)

## 3. Summarize Total PositionsMerged and PositionsSplit
total_merged = positions_merge['amount'].sum()
total_split = positions_split['amount'].sum()

print(f"Total PositionsMerged Amount: {total_merged}")
print(f"Total PositionsSplit Amount: {total_split}")

## 4. Group by `conditionId`
merge_by_condition = positions_merge.groupby('conditionId')['amount'].sum().reset_index().rename(columns={'amount': 'total_merged'})
split_by_condition = positions_split.groupby('conditionId')['amount'].sum().reset_index().rename(columns={'amount': 'total_split'})

print("PositionsMerged by conditionId:")
display(merge_by_condition)

print("\nPositionsSplit by conditionId:")
display(split_by_condition)

## 5. Merge the Grouped DataFrames and Calculate Difference
merged_data = pd.merge(split_by_condition, merge_by_condition, on='conditionId', how='outer')
merged_data['total_split'] = merged_data['total_split'].fillna(0)
merged_data['total_merged'] = merged_data['total_merged'].fillna(0)
merged_data['split_minus_merge'] = merged_data['total_split'] - merged_data['total_merged']

print("Merged Data with Differences:")
display(merged_data)

## 6. Top 5 `conditionId`s Based on Difference
top_5 = merged_data.sort_values(by='split_minus_merge', ascending=False).head(5)
print("Top 5 conditionIds by (PositionsSplit - PositionsMerge):")
display(top_5)

## 7. Calculate Total Liquidity
total_liquidity = total_split + total_merged
print(f"Total Liquidity: {total_liquidity}")

## 8. Summary
print(f"Total PositionsMerged Amount: {total_merged}")
print(f"Total PositionsSplit Amount: {total_split}")
print(f"Total Liquidity: {total_liquidity}")

print("\nTop 5 conditionIds by (PositionsSplit - PositionsMerge):")
display(top_5)
```

```python
## 9. Load Market Data and Merge
# Load the all_market_data.json file
import json

# Assuming 'all_market_data.json' is in the same directory
with open('gamma-market/all_market_data.json', 'r') as f:
    market_data_json = json.load(f)

# Convert the JSON data into a DataFrame
market_data = pd.DataFrame(market_data_json)

# Extract relevant columns
market_info = market_data[['conditionId', 'question', 'volume', 'outcomes']]

# Ensure conditionId is of type string
market_info['conditionId'] = market_info['conditionId'].astype(str)

# Merge the market information with your merged_data DataFrame
merged_with_market = pd.merge(merged_data, market_info, on='conditionId', how='left')

# Display the merged DataFrame
print("Merged Data with Market Information:")
display(merged_with_market)
```

```python
## 10. Compare Liquidity
# Convert 'volume' from string to numeric, handling any errors
merged_with_market['volume'] = pd.to_numeric(merged_with_market['volume'], errors='coerce').fillna(0)

# Calculate the total liquidity from your calculations
merged_with_market['calculated_liquidity'] = merged_with_market['total_split'] + merged_with_market['total_merged']

# Compare the liquidity from JSON vs. calculated
merged_with_market['liquidity_difference'] = merged_with_market['volume'] - merged_with_market['calculated_liquidity']

# Display the DataFrame with the liquidity comparison
print("Liquidity Comparison:")
display(merged_with_market[['conditionId', 'question', 'volume', 'calculated_liquidity', 'liquidity_difference']])
```

```python
## 11. Top 5 ConditionIds with Market Info
# Get the top 5 based on 'split_minus_merge'
top_5_with_market = merged_with_market.sort_values(by='split_minus_merge', ascending=False).head(5)

print("Top 5 conditionIds with Market Information:")
display(top_5_with_market[['conditionId', 'question', 'total_split', 'total_merged', 'split_minus_merge', 'volume', 'calculated_liquidity', 'liquidity_difference']])
```
