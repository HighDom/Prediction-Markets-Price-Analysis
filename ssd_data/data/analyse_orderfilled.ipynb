```python
import pandas as pd
from tqdm import tqdm

start = 35
end = 61

# Read the CSV file
df = pd.read_csv(f'orderfilled_events/orderfilled_{start}-{end}_no_ctfexchange.csv')

# Initialize total USDC traded
total_usdc_traded = 0

# List to store trades involving USDC
usdc_trades = []

# Lowest Number Block and Highest Number Block
lowest_blockNumber = df['blockNumber'].min()
highest_blockNumber = df['blockNumber'].max()

# Iterate over each row in the DataFrame
for index, row in tqdm(df.iterrows(), total=1053820):
    
    usdc_amount = 0
    trade_details = {}

    # Check if makerAssetId is USDC (0)

    if (row['makerAssetId'] == '0'):
        usdc_amount += row['makerAmountFilled']
        total_usdc_traded += row['makerAmountFilled']

    # Check if takerAssetId is USDC (0)
    if row['takerAssetId'] == '0':
        usdc_amount += row['takerAmountFilled']
        total_usdc_traded += row['takerAmountFilled']

    # If USDC was traded in this row, add to the list
    if usdc_amount > 0:
        trade_details = {
            'event': row['event'],
            'blockNumber': row['blockNumber'],
            'txIndex': row['txIndex'],
            'orderHash': row['orderHash'],
            'usdc_amount': usdc_amount
        }
        usdc_trades.append(trade_details)


```

```python
class bcolors:
    HEADER = '\033[95m'
    OKBLUE = '\033[94m'
    OKCYAN = '\033[96m'
    OKGREEN = '\033[92m'
    WARNING = '\033[93m'
    FAIL = '\033[91m'
    ENDC = '\033[0m'
    BOLD = '\033[1m'
    UNDERLINE = '\033[4m'
# Convert total USDC traded to USD
total_usdc_traded_usd = total_usdc_traded / 1_000_000

print(f"Total USDC traded: ${total_usdc_traded_usd}")

# Sort the trades by USDC amount in descending order
usdc_trades_sorted = sorted(usdc_trades, key=lambda x: x['usdc_amount'], reverse=True)

print(f"\n Lowest BlockNumber {bcolors.WARNING}{lowest_blockNumber}{bcolors.ENDC}\t Nov-28-2022 <- 36183855")
print(f"\n Highest BlockNumber {bcolors.WARNING}{highest_blockNumber}{bcolors.ENDC} \t Aug-24-2024 <- 60999989")

# Get the top 10 largest trades
top_10_trades = usdc_trades_sorted[:10]

print("\nTop 10 largest trades by USDC amount:")
for trade in top_10_trades:
    usdc_amount_usd = trade['usdc_amount'] / 1_000_000
    print(f"BlockNumber: {trade['blockNumber']}, txIndex: {trade['txIndex']} , USDC Amount: {bcolors.OKCYAN}${usdc_amount_usd}{bcolors.ENDC}")
```
