```python
import csv
from web3 import Web3

# Define event signatures and compute their keccak256 hashes
event_signatures = {
    'OrderFilled': 'OrderFilled(bytes32,address,address,uint256,uint256,uint256,uint256,uint256)',
    'OrdersMatched': 'OrdersMatched(bytes32,address,uint256,uint256,uint256,uint256)',
    'PositionsMerge': 'PositionsMerge(address,address,bytes32,bytes32,uint256[],uint256)',
    'PositionsMergeSimplified': 'PositionsMerge(address,bytes32,uint256)',
    'PositionSplit': 'PositionSplit(address,address,bytes32,bytes32,uint256[],uint256)',
    'PositionSplitSimplified': 'PositionSplit(address,bytes32,uint256)',
    'TokenRegistered': 'TokenRegistered(uint256,uint256,bytes32)',
    'NewOperator': 'NewOperator(address,address)',
    'FeeCharged': 'FeeCharged(address,uint256,uint256)',
    'NewAdmin': 'NewAdmin(address,address)',
    'TransferSingle': 'TransferSingle(address,address,address,uint256,uint256)',
    'TransferBatch': 'TransferBatch(address,address,address,uint256[],uint256[])'
}

# Compute keccak256 hashes for event signatures
event_topic0 = {name: Web3.keccak(text=signature).hex() for name, signature in event_signatures.items()}
print("Event topic0 hashes:")
for event_name, hash_value in event_topic0.items():
    print(f"{event_name}: {hash_value}")

# Specify input and output CSV filenames
input_filename = 'ctf_exchange_neg_risk/logs_52M.csv'
output_filename = 'ctf_exchange_neg_risk/logs_52M_decoded.csv'

# Define the CSV fieldnames
fieldnames = ['blockNumber', 'txIndex', 'logIndex', 'removed', 'topic0', 'topic1', 'topic2', 'topic3', 'data']

# Read the CSV data, add the 'event' column, and write to a new CSV file
with open(input_filename, 'r') as csvfile_in, open(output_filename, 'w', newline='') as csvfile_out:
    reader = csv.DictReader(csvfile_in, fieldnames=fieldnames)
    writer_fieldnames = ['event'] + fieldnames
    writer = csv.DictWriter(csvfile_out, fieldnames=writer_fieldnames)
    writer.writeheader()
    for row in reader:
        topic0 = row['topic0'].lower()
        event_name = 'UNKNOWN_EVENT'
        for name, hash_value in event_topic0.items():
            if topic0 == '0x' + hash_value.lower():
                event_name = name
                break
        row_with_event = {'event': event_name}
        row_with_event.update(row)
        writer.writerow(row_with_event)

print(f"Processed logs have been written to '{output_filename}'.")
```
